# 01 — Data Validation

اعتبارسنجی فایل‌های خام SST و SPI و بررسی سال‌ها، فصل‌ها، مناطق و مقادیر گمشده.

```python
from pathlib import Path
import pandas as pd
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
RAW = ROOT / 'data' / 'raw'
```

```python
sst_path = RAW / "sst_source.xlsx"
spi_path = RAW / "spi_source.xlsx"

print("SST sheets:", pd.ExcelFile(sst_path).sheet_names)
print("SPI sheets:", pd.ExcelFile(spi_path).sheet_names)
```

```python
# SST: 133 observations per season
for sheet in pd.ExcelFile(sst_path).sheet_names:
    raw = pd.read_excel(sst_path, sheet_name=sheet, header=None)
    data = raw.iloc[2:].dropna(how="all")
    print(sheet, data.shape)
```

```python
# SPI: 133 years × 154 regions per season
for sheet in pd.ExcelFile(spi_path).sheet_names:
    raw = pd.read_excel(spi_path, sheet_name=sheet, header=None)
    print(sheet, raw.shape)
    print("years:", raw.iloc[:, 1].dropna().iloc[0], "to", raw.iloc[:, 1].dropna().iloc[-1])
```
