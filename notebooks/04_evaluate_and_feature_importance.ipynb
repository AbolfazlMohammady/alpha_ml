# 04 — Evaluation & Feature Importance

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
METRICS = ROOT / "outputs" / "metrics"
```

```python
metrics = pd.read_csv(METRICS / "model_metrics.csv")
importance = pd.read_csv(METRICS / "feature_importance.csv")
predictions = pd.read_csv(METRICS / "predictions.csv")

display(metrics.describe())
display(metrics.groupby("current_season")[["r2","rmse","mae"]].mean())
```

```python
feature_summary = (
    importance.groupby("feature", as_index=False)["importance"]
    .mean()
    .sort_values("importance", ascending=False)
)
display(feature_summary)
```

```python
plt.figure(figsize=(10, 5))
plt.barh(feature_summary["feature"], feature_summary["importance"])
plt.gca().invert_yaxis()
plt.xlabel("Mean Random Forest Feature Importance")
plt.title("Average Feature Importance Across Region/Season Models")
plt.tight_layout()
plt.show()
```

```python
sample = predictions.query("region_id == 1 and current_season == 'summer'")
plt.figure(figsize=(10, 5))
plt.plot(sample["target_year"], sample["actual_spi"], label="Actual")
plt.plot(sample["target_year"], sample["predicted_spi"], label="Predicted")
plt.xlabel("Target year")
plt.ylabel("SPI")
plt.title("Region 001 — Summer to Autumn: Actual vs Predicted")
plt.legend()
plt.tight_layout()
plt.show()
```
